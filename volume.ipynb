```python
import requests
from dotenv import load_dotenv
import os
import json
import pandas as pd

load_dotenv()
```

```python
volumes = pd.read_csv("volumes.csv")
volumes.head()
```

```python
batman_volume=volumes['api_detail_url'][0]
batman_volume
```

```python
API_KEY = os.getenv("API_KEY")
BASE_URL = "https://comicvine.gamespot.com/api/"
endpoint = "volumes/"
url = batman_volume

headers = {
    "User-Agent": "absolutePipeline"  
}

params = {
        "api_key": API_KEY,
        "format": "json", 
        "field_list": "issues"
}

response = requests.get(url=url, params=params, headers=headers)

response.json()
```

```python
data = response.json()

with open("output.json", "w") as f:
    json.dump(data, f, indent=2)
```

```python
issues = data["results"]["issues"]
df = pd.DataFrame(issues)
df
```

```python

url = df['api_detail_url'][18]

headers = {
    "User-Agent": "absolutePipeline"  
}

params = {
        "api_key": API_KEY,
        "format": "json", 
        "field_list": ["date_last_updated,character_credits,description,id,image,object_credits,person_credits,store_date,volume"]
}

response = requests.get(url=url, params=params, headers=headers)

response.json()
```

```python
data = response.json()

with open("issue_details.json", "w") as f:
    json.dump(data, f, indent=2)
```

```python
#Get characters in issue
characters_json = data["results"]["character_credits"]
characters_list = []
for character in characters_json: 
    characters_list.append(character.get("name"))
    #print(character)
# get description
description= data["results"]["description"]
# get image
image_url = data["results"]["image"]["medium_url"]
# get objects
objects_json = data["results"]["object_credits"]
objects_list = []
for object in objects_json: 
    objects_list.append(object.get("name"))
# get credits
persons_json = data["results"]["person_credits"] 
for person in persons_json: 
    colorists = [p["name"] for p in persons_json if "colorist" in p.get("role", "")]
    artists = [p["name"] for p in persons_json if "artist" in p.get("role", "")]
    writers = [p["name"] for p in persons_json if "writer" in p.get("role", "")]
#  get store date
store_date = data["results"]["store_date"] 
# volume id
volume_id = data["results"]["volume"]["id"]
```

```python
df.at[18, "characters"] = characters_list
df.at[18, "description"] = description
df.at[18, "image_url"] = image_url
df.at[18, "objects"] = objects_list
df.at[18, "colorists"] = colorists
df.at[18, "artists"] = artists
df.at[18, "writers"] = writers
df.at[18, "store_date"] = store_date
df.at[18, "volume_id"] = int(volume_id)
df
```
